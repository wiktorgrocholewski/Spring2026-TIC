```python
# US Corn Belt

LOCATIONS_US = {
    "central_iowa": (42.0, -93.5),        # Heart of corn/soy production
    "central_illinois": (40.1, -89.4),     # Major corn/soy state
    "central_indiana": (40.0, -86.2),      # Eastern corn belt
}

# Brazil soy regions (optional, for soybean-specific weather)
LOCATIONS_BRAZIL = {
    "mato_grosso": (-12.6, -55.5),
    "parana":      (-24.0, -51.5),
    "rio_grande_sul": (-29.5, -53.0),   # <-- add this
}

# US wheat belt locations
LOCATIONS_WHEAT = {
    "kansas":       (38.5, -98.0),    # biggest HRW wheat state
    "oklahoma":     (35.5, -97.5),    # second biggest HRW
    "north_dakota": (47.0, -100.0),   # spring wheat
}
# split API call into 2 to prevent error from requesting too much data at once
# Call 1: corn/soy locations (existing)
weather_corn_soy = fetch_weather_data(
    locations={**LOCATIONS_US, **LOCATIONS_BRAZIL},
    start_date="2005-01-01",
    save_path="../data/weather_corn_soy.csv",
)
```

```python
# Call 2: wheat locations only
weather_wheat = fetch_weather_data(
    locations=LOCATIONS_WHEAT,
    start_date="2005-01-01",
    save_path="../data/weather_wheat.csv",
)

# Then merge both weather files
import pandas as pd
weather_combined = pd.concat([weather_corn_soy, weather_wheat], axis=1)
weather_combined.to_csv("../data/weather_combined.csv")
```
